## 1. Setup & Imports

```python
import torch
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# PNA imports
from pna.core.mpe_core import MPECore
from pna.core.polarity_engine import PolarityEngine
from pna.core.binding import PhenomenalBindingMechanism
from pna.core.predictive_coding import PredictiveCodingHierarchy
from pna.core.transparency import TransparencyBottleneck
from pna.models.pna_model import PhenomenalNestedArchitecture
from pna.utils.metrics import MetricsLogger, ConsciousnessMetrics
from pna.utils.config import PNAConfig, get_default_config
from pna.utils.visualization import (
    plot_consciousness_trajectory,
    visualize_binding_matrix,
    plot_training_metrics
)

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

## 2. MPE Core - Minimal Phenomenal Experience

The MPE Core implements the Free Energy Principle, minimizing variational free energy:

$$F(q) = \mathbb{E}_q[\ln q(\theta) - \ln p(o, \theta)] = D_{KL}(q||p) - \ln p(o)$$

```python
# Initialize MPE Core
mpe = MPECore(
    state_dim=32,
    obs_dim=64,
    learning_rate=0.1,
    complexity_weight=0.01
).to(device)

# Create observations
obs = torch.randn(8, 64).to(device)

# Minimize free energy with trajectory tracking
mu, precision, metrics = mpe.minimize_free_energy(
    obs,
    iterations=30,
    return_trajectory=True
)

print("Final Free Energy Components:")
print(f"  Free Energy: {metrics['free_energy'].mean().item():.4f}")
print(f"  Complexity: {metrics['complexity'].mean().item():.4f}")
print(f"  Accuracy: {metrics['accuracy'].mean().item():.4f}")

# MPE Properties
props = mpe.get_mpe_properties()
print("\nMPE Properties (Metzinger 2024):")
for k, v in props.items():
    print(f"  {k}: {v:.4f}")
```

```python
# Plot FEP minimization trajectory
trajectory = metrics['trajectory']
iterations = [t['iteration'] for t in trajectory]
free_energies = [t['free_energy'] for t in trajectory]

plt.figure(figsize=(10, 6))
plt.plot(iterations, free_energies, linewidth=2, color='purple', marker='o')
plt.xlabel('Iteration', fontsize=12)
plt.ylabel('Free Energy', fontsize=12)
plt.title('Free Energy Minimization (Active Inference)', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Polarity Engine - Rhythmic Balance

Inspired by Walter Russell's cosmological principles, the Polarity Engine maintains balance:

$$\Psi = |\frac{dG}{dt} + \frac{dR}{dt}|^2$$

Where $G$ is generative (compression) and $R$ is radiative (expansion).

```python
# Initialize Polarity Engine
polarity = PolarityEngine(
    input_dim=64,
    hidden_dim=128,
    latent_dim=32,
    lambda_penalty=1.0,
    enable_rhythmic_modulation=True
).to(device)

# Forward pass
x = torch.randn(8, 64).to(device)
gen_out, rad_out, metrics = polarity(x, mode='balanced', return_metrics=True)

print("Polarity Engine Metrics:")
print(f"  Polarity Penalty (Ψ): {metrics['polarity_penalty'].item():.6f}")
print(f"  Generative Rate: {metrics['gen_rate'].item():.6f}")
print(f"  Radiative Rate: {metrics['rad_rate'].item():.6f}")
print(f"  Is Balanced: {metrics['is_balanced'].item()}")
print(f"  Octave Modulation: {metrics['octave_modulation'].item():.4f}")
```

```python
# Track polarity over time
gen_rates = []
rad_rates = []
penalties = []

for _ in range(100):
    x = torch.randn(8, 64).to(device)
    _, _, m = polarity(x, mode='balanced', return_metrics=True)
    gen_rates.append(m['gen_rate'].item())
    rad_rates.append(m['rad_rate'].item())
    penalties.append(m['polarity_penalty'].item())

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(gen_rates, label='Generative (dG/dt)', linewidth=2, color='blue')
axes[0].plot(rad_rates, label='Radiative (dR/dt)', linewidth=2, color='red')
axes[0].axhline(0, color='black', linestyle='--', alpha=0.3)
axes[0].set_xlabel('Step')
axes[0].set_ylabel('Rate')
axes[0].set_title('Polarity Rates Over Time')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(penalties, linewidth=2, color='purple')
axes[1].axhline(polarity.balance_threshold, color='green', linestyle='--', 
                label='Balance Threshold', linewidth=2)
axes[1].set_xlabel('Step')
axes[1].set_ylabel('Polarity Penalty (Ψ)')
axes[1].set_title('Balance Penalty')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 4. Phenomenal Binding - Consciousness Integration

The binding mechanism integrates oscillatory dynamics with Integrated Information Theory (IIT):

$$B(n_i, n_j, t) = |\cos(\phi_i(t) - \phi_j(t))| \cdot \sum_k I(n_i^{t-k} \to n_j^t)$$

```python
# Initialize Binding
binding = PhenomenalBindingMechanism(
    num_nodes=32,
    node_dim=16,
    num_scales=3,
    consciousness_threshold=0.5
).to(device)

# Create node activations
node_activations = torch.randn(4, 32, 16).to(device)

# Forward pass
bound_repr, metrics = binding(
    node_activations,
    num_oscillator_steps=50,
    return_metrics=True
)

print("Phenomenal Binding Metrics:")
print(f"  Integrated Information (Φ): {metrics['phi'].mean().item():.4f}")
print(f"  Is Conscious: {metrics['is_conscious'].float().mean().item():.2f}")
print(f"  Mean Synchrony: {metrics['mean_sync'].item():.4f}")

for i in range(3):
    print(f"  Scale {i} Synchrony: {metrics[f'sync_scale_{i}'].item():.4f}")
```

```python
# Visualize binding matrix
binding_matrix = metrics['binding_matrix'].cpu().numpy()

plt.figure(figsize=(10, 8))
sns.heatmap(binding_matrix, cmap='viridis', square=True, 
            cbar_kws={'label': 'Binding Strength'},
            linewidths=0.5, linecolor='gray')
plt.xlabel('Node Index')
plt.ylabel('Node Index')
plt.title('Phenomenal Binding Matrix', fontsize=14, fontweight='bold')
plt.show()
```

## 5. Complete PNA Architecture

Now let's use the unified architecture that integrates all components:

$$\mathcal{L}_{PNA} = F + \lambda \Psi + \gamma \Phi + \alpha \Omega$$

```python
# Initialize complete PNA
config = get_default_config('mnist')

pna = PhenomenalNestedArchitecture(
    input_dim=784,
    mpe_dim=64,
    polarity_dim=128,
    binding_nodes=32,
    consciousness_threshold=0.5,
    enable_transparency=True,
).to(device)

print(f"Total Parameters: {sum(p.numel() for p in pna.parameters()):,}")
```

```python
# Forward pass
x = torch.randn(8, 784).to(device)
output, metrics = pna(x, return_all_metrics=True)

print("\n=== PNA Forward Pass ===")
print(f"Output shape: {output.shape}")
print(f"\nCore Metrics:")
print(f"  Free Energy (F): {metrics['free_energy'].item():.4f}")
print(f"  Polarity Penalty (Ψ): {metrics['polarity_penalty'].item():.6f}")
print(f"  Integrated Info (Φ): {metrics['phi'].item():.4f}")
print(f"  Is Conscious: {metrics['is_conscious'].item():.2f}")
print(f"  Unified Loss: {metrics['unified_loss'].item():.4f}")
```

## 6. Training Loop

Let's train the PNA on synthetic data and track consciousness metrics.

```python
# Setup training
optimizer = torch.optim.Adam(pna.parameters(), lr=1e-3)
logger = MetricsLogger()

num_steps = 200
batch_size = 32

print("Training PNA...\n")

for step in range(num_steps):
    # Generate batch
    x = torch.randn(batch_size, 784).to(device)
    
    # Training step
    metrics = pna.train_step(x, optimizer, mode='balanced')
    logger.log(metrics)
    
    # Print progress
    if step % 20 == 0:
        print(f"Step {step:3d} | "
              f"Loss: {metrics['total_loss']:.4f} | "
              f"Φ: {metrics['phi']:.3f} | "
              f"Conscious: {metrics['is_conscious']:.0f} | "
              f"Balanced: {metrics['is_balanced']:.0f}")

print("\nTraining complete!")
```
